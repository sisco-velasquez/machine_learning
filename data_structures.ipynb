```python
class Node:
    def __init__(self, data=None, next=None):
        self.data = data
        self.next = next

class LinkedList:
    def __init__(self):
        self.head = None

    def insert_in_the_beginning(self, data):
        node = Node(data, self.head)
        self.head = node

    def print(self):
        if self.head is None:
            print("linked list is empty")
            return

        itr = self.head
        llstr = ''

        while itr:
            llstr += str(itr.data) + '-->'
            itr = itr.next

        print(llstr)

    def insert_at_end(self, data):
        if self.head is None:
            self.head = Node(data, None)
            return

        itr = self.head
        while itr.next:
            itr = itr.next

        itr.next = Node(data, None)

    # --- New Functions ---

    def get_length(self):
        count = 0
        itr = self.head
        while itr:
            count += 1
            itr = itr.next
        return count

    def insert_at(self, index, data):
        if index < 0 or index > self.get_length():
            raise Exception("Invalid Index")

        if index == 0:
            self.insert_in_the_beginning(data)
            return

        count = 0
        itr = self.head
        while itr:
            if count == index - 1:
                node = Node(data, itr.next)
                itr.next = node
                break

            itr = itr.next
            count += 1

    def remove_at(self, index):
        if index < 0 or index >= self.get_length():
            raise Exception("Invalid Index")

        if index == 0:
            self.head = self.head.next
            return

        count = 0
        itr = self.head
        while itr:
            if count == index - 1:
                itr.next = itr.next.next
                break

            itr = itr.next
            count += 1

    def insert_values(self, data_list):
        self.head = None
        for data in data_list:
            self.insert_at_end(data)
    def insert_after_value(self, data_after, data_to_insert):
        if self.head is None:
            return

        itr = self.head
        while itr:
            if itr.data == data_after:
                node = Node(data_to_insert, itr.next)
                itr.next = Node(data_to_insert, itr.next)
                break  # Remove this break if you want to insert after ALL occurrences

            itr = itr.next

if __name__ == '__main__':
    ll = LinkedList()

    # Bulk insert
    ll.insert_values(["Apple", "Banana", "Cherry", "Date"])
    ll.print()

    print("Length:", ll.get_length())

    # Remove item at index 2 (Cherry)
    ll.remove_at(2)
    ll.print()

    # Insert at index 1
    ll.insert_at(2, "Blueberry")
    ll.print()

    ll.insert_after_value("Apple", "Kiwi")
    ll.print()
```

### Hash Table

```python
class HashTable:
    def __init__(self, size=10):
        self.size = size
        # Initialize the array with empty lists to handle chaining
        self.arr = [[] for _ in range(self.size)]

    def get_hash(self, key):
        # Simple hash function: sum of ASCII values modulo array size
        hash_total = sum(ord(char) for char in str(key))
        return hash_total % self.size

    def __setitem__(self, key, val):
        h = self.get_hash(key)

        # Check if key already exists in the chain to update it
        found = False
        for index, element in enumerate(self.arr[h]):
            if len(element) == 2 and element[0] == key:
                self.arr[h][index] = (key, val)
                found = True
                break

        # If the key doesn't exist, append it to the chain
        if not found:
            self.arr[h].append((key, val))

    def __getitem__(self, key):
        h = self.get_hash(key)

        # Search through the chain at the calculated index
        for element in self.arr[h]:
            if element[0] == key:
                return element[1]

        raise KeyError(f"Key '{key}' not found")

    def __delitem__(self, key):
        h = self.get_hash(key)

        for index, element in enumerate(self.arr[h]):
            if element[0] == key:
                del self.arr[h][index]
                return

        raise KeyError(f"Key '{key}' not found")

    def print_table(self):
        for i, val in enumerate(self.arr):
            print(f"Bucket {i}: {val}")

if __name__ == '__main__':
    t = HashTable(size=10)

    # Insert values
    t["apple"] = 150
    t["banana"] = 200
    t["cherry"] = 50

    # 'apple' and 'cherry' might hash to the same bucket depending on the size.
    # Let's force a collision manually by finding two strings with the same hash
    # For a size 10 array, "march 6" and "march 17" both hash to 9
    t["march 6"] = 310
    t["march 17"] = 420
    t["FRANCIS"]= 312

    # Print the table to see the collision chaining in Bucket 9
    t.print_table()

    # Retrieve a value
    print("\nValue for 'march 6':", t["march 6"])

    # Delete a value
    del t["banana"]
    print("\nTable after deleting 'banana':")
    t.print_table()
```

### stack

```python
from collections import deque

class Stack:
    def __init__(self):
        self.container = deque()

    def is_empty(self):
        return len(self.container) == 0

    def push(self, item):
        # appends to the right side (the "top" of the stack)
        self.container.append(item)

    def pop(self):
        if self.is_empty():
            raise IndexError("pop from empty stack")
        # pops from the right side
        return self.container.pop()

    def peek(self):
        if self.is_empty():
            raise IndexError("peek from empty stack")
        # access the right-most item
        return self.container[-1]

    def size(self):
        return len(self.container)

    def __str__(self):
        return "Bottom -> " + " | ".join(str(item) for item in self.container) + " <- Top"

if __name__ == '__main__':
    # Initialize the deque-based stack
    s = Stack()

    # Push items
    s.push("Kali Linux")
    s.push("Windows 10 LTSC")
    s.push("Windows 11")

    print(s)

    # Peek and Pop
    print("Top item is:", s.peek())
    print("Removed:", s.pop())

    print("\nAfter popping:")
    print(s)
```

reversing a string using deque


```python
from collections import deque

class Stack:
    def __init__(self):
        self.container = deque()

    def is_empty(self):
        return len(self.container) == 0

    def push(self, item):
        self.container.append(item)

    def pop(self):
        if self.is_empty():
            raise IndexError("pop from empty stack")
        return self.container.pop()

    def size(self):
        return len(self.container)

def reverse_string(text):
    stack = Stack()

    # Step 1: Push every character in the string onto the stack
    for char in text:
        stack.push(char)

    # Step 2: Pop them off one by one and append them to a new string
    reversed_text = ""
    while not stack.is_empty():
        reversed_text += stack.pop()

    return reversed_text

if __name__ == '__main__':
    original_str = "FastAPI and PostgreSQL"

    print(f"Original string: {original_str}")
    print(f"Reversed string: {reverse_string(original_str)}")
```

queue


```python
from collections import deque

class Queue:
    def __init__(self):
        self.buffer = deque()

    def is_empty(self):
        return len(self.buffer) == 0

    def enqueue(self, item):
        # Adds the item to the back (right side) of the queue
        self.buffer.append(item)

    def dequeue(self):
        if self.is_empty():
            raise IndexError("dequeue from empty queue")
        # Removes and returns the item from the front (left side) of the queue
        return self.buffer.popleft()

    def peek(self):
        if self.is_empty():
            raise IndexError("peek from empty queue")
        # Looks at the front-most item without removing it
        return self.buffer[0]

    def size(self):
        return len(self.buffer)

    def __str__(self):
        return "Front -> " + " | ".join(str(item) for item in self.buffer) + " <- Back"

if __name__ == '__main__':
    # Let's simulate a print job queue for an office
    print_queue = Queue()

    print("--- Submitting Print Jobs (Enqueue) ---")
    print_queue.enqueue("Financial_Report.pdf")
    print_queue.enqueue("Meeting_Agenda.docx")
    print_queue.enqueue("Lunch_Menu.png")

    print(print_queue)
    print("Next job to print:", print_queue.peek())

    print("\n--- Printer Processing Jobs (Dequeue) ---")
    completed_job = print_queue.dequeue()
    print(f"Finished printing: {completed_job}")

    print("\n--- Current Queue State ---")
    print(print_queue)
    print("Jobs remaining:", print_queue.size())
```

```python
import threading
import time
from collections import deque

class Queue:
    def __init__(self):
        self.buffer = deque()

    def is_empty(self):
        return len(self.buffer) == 0

    def enqueue(self, item):
        self.buffer.append(item)

    def dequeue(self):
        if self.is_empty():
            raise IndexError("dequeue from empty queue")
        return self.buffer.popleft()

# Shared queue resource
food_queue = Queue()

def place_order(orders):
    for order in orders:
        print(f"[CASHIER] Taking order for: {order}")
        food_queue.enqueue(order)
        # Simulate the time it takes to place an order (fast)
        time.sleep(0.5)
    print("[CASHIER] All orders have been taken.")

def serve_order():
    # Wait briefly to ensure the cashier gets the first order in before the kitchen checks
    time.sleep(1)

    while True:
        if not food_queue.is_empty():
            order = food_queue.dequeue()
            print(f"   [KITCHEN] Serving: {order}")
            # Simulate the time it takes to cook and serve (slow)
            time.sleep(2)
        else:
            # If the queue is empty, the kitchen is done
            print("   [KITCHEN] All orders served. Kitchen closing.")
            break

if __name__ == '__main__':
    orders = ['Pizza', 'Samosa', 'Pasta', 'Biryani', 'Burger']

    # Define the threads
    # target defines which function the thread will run
    # args passes the arguments to that function (must be a tuple)
    t1 = threading.Thread(target=place_order, args=(orders,))
    t2 = threading.Thread(target=serve_order)

    # Start both threads so they run simultaneously
    t1.start()
    t2.start()

    # .join() tells the main program to wait here until both threads finish
    t1.join()
    t2.join()

    print("[SYSTEM] Restaurant operations completed.")
```

general trees

```python
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.children = []
        self.parent = None  # Helpful for calculating levels

    def add_child(self, child):
        # When adding a child, tell the child who its parent is
        child.parent = self
        # Then add the child to the list
        self.children.append(child)

    def get_level(self):
        level = 0
        p = self.parent
        # Traverse upwards until we hit the root (which has no parent)
        while p:
            level += 1
            p = p.parent
        return level

    def print_tree(self):
        # Calculate indentation based on depth level
        spaces = ' ' * self.get_level() * 3
        prefix = spaces + "|__" if self.parent else ""

        print(prefix + self.data)

        # Recursively print all children
        if self.children:
            for child in self.children:
                child.print_tree()

if __name__ == '__main__':
    # 1. Build the Root
    root = TreeNode("Electronics")

    # 2. Build the Level 1 Categories
    laptop = TreeNode("Laptops")
    phone = TreeNode("Cell Phones")
    tv = TreeNode("Televisions")

    # 3. Build Level 2 Sub-categories (Leaves)
    laptop.add_child(TreeNode("MacBook"))
    laptop.add_child(TreeNode("ThinkPad"))

    phone.add_child(TreeNode("iPhone"))
    phone.add_child(TreeNode("Samsung"))

    tv.add_child(TreeNode("LG OLED"))

    # 4. Connect everything to the Root
    root.add_child(laptop)
    root.add_child(phone)
    root.add_child(tv)

    # 5. Print the visual structure
    root.print_tree()
```

binary tree

```python
class BinarySearchTreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

    def add_child(self, data):
        if data == self.data:
            return  # BSTs typically do not allow duplicate values

        if data < self.data:
            # Go left
            if self.left:
                self.left.add_child(data)
            else:
                self.left = BinarySearchTreeNode(data)
        else:
            # Go right
            if self.right:
                self.right.add_child(data)
            else:
                self.right = BinarySearchTreeNode(data)

    def search(self, val):
        if self.data == val:
            return True

        if val < self.data:
            # Value might be in the left subtree
            if self.left:
                return self.left.search(val)
            else:
                return False

        if val > self.data:
            # Value might be in the right subtree
            if self.right:
                return self.right.search(val)
            else:
                return False

    def in_order_traversal(self):
        elements = []

        # 1. Visit Left Tree
        if self.left:
            elements += self.left.in_order_traversal()

        # 2. Visit Base Node
        elements.append(self.data)

        # 3. Visit Right Tree
        if self.right:
            elements += self.right.in_order_traversal()

        return elements

def build_tree(elements):
    print("Building tree with these elements:", elements)
    root = BinarySearchTreeNode(elements[0])
    for i in range(1, len(elements)):
        root.add_child(elements[i])
    return root

if __name__ == '__main__':
    # Insert unsorted numbers
    numbers = [17, 4, 1, 20, 9, 23, 18, 34]

    bst = build_tree(numbers)

    # In-order traversal naturally sorts the output
    print("In-order traversal:", bst.in_order_traversal())

    # Fast searching
    print("Is 20 in the tree?", bst.search(20))
    print("Is 100 in the tree?", bst.search(100))
```
