### Arrays

```python
expenses = [
    {"jan": "2200"},
    {"feb": "2350"},
    {"mar": "2600"},
    {"apr": "2130"},
    {"may": "2190"}
]

# Extract the values and convert them to integers
jan_expense = int(expenses[0]["jan"])
feb_expense = int(expenses[1]["feb"])

# Calculate the difference
difference = feb_expense - jan_expense
print(f"In February, you spent ${difference} extra compared to January.")
```

```python
jan_expense = int(expenses[0]["jan"])
feb_expense = int(expenses[1]["feb"])
mar_expense = int(expenses[2]["mar"])

total = jan_expense + feb_expense + mar_expense
print(f"Total expenses in the first quarter: ${total}")
```

```python
for i in range(len(expenses)):
    expense_value = list(expenses[i].values())[0]
    if expense_value == "2000":
        print(f"Found 2000 at index {i}")
        break
else:
    print("An expense of 2000 was not found.")
```

```python
expenses.insert(5, {"jun": "1980"})
print(expenses)
```

```python
# To remove an item by index, use .pop() with parentheses instead of .remove[]
expenses.pop(6)
print(expenses)
```

```python
# Approach 1: Replace the entire dictionary at a specific index (e.g., January at index 0)
expenses[3] = {"apr": "1930"}
print("After updating April's dictionary:", expenses)

```

####linked list

```python
class Node:
    def __init__(self, data=None, next=None):
        self.data = data
        self.next = next

class LinkedList:
    def __init__(self):
        self.head = None

    def insert_in_the_beginning(self, data):
        node = Node(data, self.head)
        self.head = node

    def print(self):
        if self.head is None:
            print("linked list is empty")
            return

        itr = self.head
        llstr = ''

        while itr:
            llstr += str(itr.data) + '-->'
            itr = itr.next

        print(llstr)

    def insert_at_end(self, data):
        if self.head is None:
            self.head = Node(data, None)
            return

        itr = self.head
        while itr.next:
            itr = itr.next

        itr.next = Node(data, None)

    # --- New Functions ---

    def get_length(self):
        count = 0
        itr = self.head
        while itr:
            count += 1
            itr = itr.next
        return count

    def insert_at(self, index, data):
        if index < 0 or index > self.get_length():
            raise Exception("Invalid Index")

        if index == 0:
            self.insert_in_the_beginning(data)
            return

        count = 0
        itr = self.head
        while itr:
            if count == index - 1:
                node = Node(data, itr.next)
                itr.next = node
                break

            itr = itr.next
            count += 1

    def remove_at(self, index):
        if index < 0 or index >= self.get_length():
            raise Exception("Invalid Index")

        if index == 0:
            self.head = self.head.next
            return

        count = 0
        itr = self.head
        while itr:
            if count == index - 1:
                itr.next = itr.next.next
                break

            itr = itr.next
            count += 1

    def insert_values(self, data_list):
        self.head = None
        for data in data_list:
            self.insert_at_end(data)

if __name__ == '__main__':
    ll = LinkedList()

    # Bulk insert
    ll.insert_values(["Apple", "Banana", "Cherry", "Date"])
    ll.print()

    print("Length:", ll.get_length())

    # Remove item at index 2 (Cherry)
    ll.remove_at(2)
    ll.print()

    # Insert at index 1
    ll.insert_at(1, "Blueberry")
    ll.print()
```
